##Phishing problem:
Problem Statement: Phishing Website Detection using Machine Learning

In the digital age, online security has become a major concern as users increasingly rely on the internet for financial transactions, communication, and information exchange. Phishing attacks are one of the most common and dangerous forms of cybercrime, where attackers create fraudulent websites that mimic legitimate ones to steal sensitive information such as usernames, passwords, and credit card details.

Traditional methods of detecting phishing websites such as maintaining blacklists or using rule based filters are often ineffective against newly created or rapidly changing phishing sites. These methods require frequent manual updates and fail to generalize well to novel attacks.

This project aims to develop a machine learning based phishing website detection model that can automatically identify whether a given website is legitimate or phishing based on its URL structure, domain features, and webpage characteristics. By learning from patterns in historical data, the model should be able to generalize to unseen websites and make real-time predictions.

##Stage 2

Want to try model tuning for better efficiency, then cross validation for further evaluation of the model

This loop iterates over a list of values for the hyperparameter C.
C is the inverse of the regularization strength in Logistic Regression.
Smaller values mean stronger regularization (more penalty on complexity).
Larger values mean weaker regularization.
Creates a Logistic Regression model for the current C value.
solver='liblinear' is an algorithm choice suitable for small datasets and binary classification.
max_iter=1000 allows up to 1000 iterations for convergence, ensuring the model can fully train.

```python
#Tuned Logistic Regression (with real cross-validation via GridSearchCV)
from sklearn.model_selection import GridSearchCV

param_grid_lr = {'C': [0.01, 0.1, 1, 10]}
grid_lr = GridSearchCV(
    LogisticRegression(max_iter=1000, solver='liblinear'),
    param_grid_lr, cv=5, scoring='accuracy'
)
grid_lr.fit(X_train_scaled, y_train)

print("Best C:", grid_lr.best_params_)
print("Best CV accuracy:", grid_lr.best_score_)

best_lr = grid_lr.best_estimator_
y_lr_train_pred = best_lr.predict(X_train_scaled)
y_lr_test_pred = best_lr.predict(X_test_scaled)

acc_train_lr_tuned = accuracy_score(y_train, y_lr_train_pred)
acc_test_lr_tuned = accuracy_score(y_test, y_lr_test_pred)
print(f"Tuned LR - Train Acc: {acc_train_lr_tuned:.2f}, Test Acc: {acc_test_lr_tuned:.2f}")

print("\nConfusion Matrix (Test):", confusion_matrix(y_test, y_lr_test_pred))
print("\nClassification Report (Test):", classification_report(y_test, y_lr_test_pred))
```

```python
#Applying FFNN
# Convert Y from -1/1 to 0/1

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import StandardScaler


import numpy as np
y_train_enc = np.where(y_train == -1, 0, 1)
y_test_enc = np.where(y_test == -1, 0, 1)


model = Sequential([
    Dense(128, input_dim=X_train_scaled.shape[1], activation='relu'),
    Dropout(0.4),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid')
])

# Compile model
model.compile(optimizer=Adam(learning_rate=0.001),
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Train
history = model.fit(X_train_scaled, y_train_enc,
                    validation_data=(X_test_scaled, y_test_enc),
                    epochs=50, batch_size=32, verbose=1)

# Evaluate
loss, acc = model.evaluate(X_test_scaled, y_test_enc)
print(f"Test Accuracy: {acc*100:.2f}%")
```

```python
#APPLYING
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, Flatten, Dense, Dropout

# Reshape for 1D CNN
X_train_cnn = X_train_scaled.reshape((X_train_scaled.shape[0], X_train_scaled.shape[1], 1))
X_test_cnn = X_test_scaled.reshape((X_test_scaled.shape[0], X_test_scaled.shape[1], 1))

cnn_model = Sequential([
    Conv1D(64, kernel_size=3, activation='relu', input_shape=(X_train_scaled.shape[1], 1)),
    Dropout(0.3),
    Conv1D(32, kernel_size=3, activation='relu'),
    Flatten(),
    Dense(32, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

cnn_model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
cnn_model.fit(X_train_cnn, y_train_enc,
              validation_data=(X_test_cnn, y_test_enc),
              epochs=50, batch_size=32, verbose=1)

loss, acc = cnn_model.evaluate(X_test_cnn, y_test_enc)
print(f"CNN Test Accuracy: {acc*100:.2f}%")
```

```python
import torch
from torch import nn
input_size = X_train.shape[1]
class AE(nn.Module):
    def __init__(self,input_dim):
        super(AE, self).__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),
            nn.Linear(64, 32),
            nn.ReLU(),
            nn.Linear(32, 16),   # bottleneck (latent space)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.Linear(16, 32),
            nn.ReLU(),
            nn.Linear(32, 64),
            nn.ReLU(),
            nn.Linear(64, input_dim),
            nn.Sigmoid()        # outputs reconstructed features
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded
X_train_t = torch.tensor(X_train.values, dtype=torch.float32)
X_test_t  = torch.tensor(X_test.values, dtype=torch.float32)

model = AE(input_dim=input_size)

train_loader = DataLoader(X_train_t, batch_size=32, shuffle=True)
```

```python
input_size = X_train.shape[1]   # 31 features

model = AE(input_dim=input_size)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

for epoch in range(100):
    for batch in train_loader:
        preds = model(batch)
        loss = criterion(preds, batch)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

    print(f"Epoch {epoch+1}: Loss = {loss.item():.4f}")

```
